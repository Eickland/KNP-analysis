```python
from HumSpectra.sample import SampleData, SampleCollection
import HumSpectra.fluorescence as fl
import HumSpectra.utilits as ut

from pathlib import Path
import numpy as np
```

```python
collection = SampleCollection(cache_dir=Path.joinpath(Path.cwd().parent.parent,"Initial_data","data_cache"))
```

```python
wdir = Path.cwd().parent.parent
fluo_path = Path.joinpath(wdir,"Initial_data","2025", "Fluorescence")
fluo_spectra = ut.load_spectra_data(fluo_path, fl.read_fluo_3d)
```

```python
sample = collection.create_sample_from_data(
        sample_id="sample_001",
        fluorescence_eem=fluo_spectra[0],
        measurement_params={
            "excitation_wavelengths": np.linspace(240, 450, 100),
            "emission_wavelengths": np.linspace(250, 600, 100),
            "integration_time": 0.1
        },
        sample_type="unknown",
        comments="Первый тестовый образец"
    )
```

```python
collection.save_sample("sample_001")
```
