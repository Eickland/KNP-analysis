```python
import pandas as pd
from pathlib import Path
import HumSpectra.utilits as ut
```

```python
fluo_paths = Path.cwd().parent/ 'Initial data'/ '2025' / 'Fluorescence'
uv_paths = Path.cwd().parent/ 'Initial data'/ '2025' / 'UV' / 'Individual spectra'

wdir_2025 = Path.cwd().parent / 'Initial data' / '2025'
carbon_data_init = pd.read_excel(Path.joinpath(wdir_2025, 'C_org.xlsx'))
geo_data_init = pd.read_excel(Path.joinpath(wdir_2025, 'GeoData.xlsx'))
```

```python
list_fluo_samples = [ut.extract_name_from_path(str(path)) for path in fluo_paths.rglob('*.csv')]
list_uv_samples = [ut.extract_name_from_path(str(path)) for path in fluo_paths.rglob('*.csv')]

carbon_samples = carbon_data_init.Sample.to_list()
geo_samples = geo_data_init.Sample.to_list()
```

```python
def find_unique_strings_ordered(*lists):
    """
    Версия, которая возвращает уникальные строки в порядке их первого появления в каждом списке.
    """
    if not lists:
        return {}
    
    # Используем эффективный алгоритм для определения уникальности
    frequency = {}
    for i, lst in enumerate(lists):
        for string in lst:
            if string not in frequency:
                frequency[string] = set()
            frequency[string].add(i)
    
    # Для каждого списка собираем уникальные строки, сохраняя порядок
    unique_strings = {}
    for i, lst in enumerate(lists):
        unique_strings[i] = []
        seen = set()  # Чтобы избежать дубликатов в результате
        for string in lst:
            if string not in seen and len(frequency[string]) == 1:
                unique_strings[i].append(string)
                seen.add(string)
    
    return unique_strings
```

```python
find_unique_strings_ordered(list_fluo_samples,list_uv_samples,carbon_samples,geo_samples)
```
