```python
import pandas as pd
from pathlib import Path
import HumSpectra.ultraviolet as Huv

wdir = Path.cwd().parent
uv_paths = Path.cwd() / "Исходные данные" / "2025"  / "UV"/ "Спектры по картам"
```

```python
list_dict = []

for windows_path in uv_paths.rglob('*.xlsx'):
    path = str(windows_path)

    df = pd.read_excel(path,sheet_name=None)

    list_dict.append(df)
```

```python
for dict in list_dict:

    for name,data in dict.items():
        
        if "K" in name or "К" in name:
        
            name = name[1:]

        data.to_csv(Path.joinpath(Path.cwd() / "Исходные данные" / "2025"  / "UV"/ "CSV спектры",f"ADOM-SL{name}.csv"),index=None)
```

```python
for dict in list_dict:

    for name,data in dict.items():

        split_name = name.split(sep='-')
        
        if "B" in split_name[1] or "В" in split_name[1]:
        
            name = split_name[0]+'-'

        data.to_csv(Path.joinpath(Path.cwd() / "Исходные данные" / "2025"  / "UV"/ "CSV спектры",f"{name}.csv"),index=None)
```
