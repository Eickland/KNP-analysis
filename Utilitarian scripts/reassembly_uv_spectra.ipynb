```python
import pandas as pd
from pathlib import Path
import HumSpectra.ultraviolet as Huv

wdir = Path.cwd().parent
uv_paths = Path.cwd().parent / 'Initial data'/'2025'/'UV'/'Grouped spectra'
```

```python
def rename(row):
    
    name = row
    split_name = name.split(sep='-')
    
    if 'B' in split_name[0]:
        return 'ADOM-SLB-'+ name
    
    elif 'K' in split_name[0] or 'К' in split_name[0]:
        return 'ADOM-SL'+ name[1:]
    
    else:
        return 'ADOM-SL'+ name
```

```python
list_dict = []

for windows_path in uv_paths.rglob('*.xlsx'):
    path = str(windows_path)

    df = pd.read_excel(path,sheet_name=None)

    list_dict.append(df)
```

```python
for dict in list_dict:

    for oldname,data in dict.items():
        name = rename(oldname)
        print(name)

        data.to_csv(Path.joinpath(Path.cwd().parent / 'Initial data'/'2025'/'UV'/'Individual spectra',f"{name}.csv"),index=None)
```
