```python
from pathlib import Path
import pandas as pd
```

```python
wdir_2025 = Path.cwd().parent / 'Initial data' / '2025'
toc_data_init = pd.read_excel(Path.joinpath(wdir_2025, 'TOC.xlsx')) # pyright: ignore[reportCallIssue]
map_data_init = pd.read_excel(Path.joinpath(wdir_2025, 'Карты.xlsx'))
```

```python
def rename(row):
    
    name = row
    split_name = name.split(sep='-')
    
    if 'B' in split_name[0]:
        return 'ADOM-SLB-'+ name  
    else:
        return 'ADOM-SL'+ name[1:]
```

```python
toc_data_init['Sample Name'] = toc_data_init['Sample Name'].apply(lambda x: rename(x))
```

```python
toc_data_init.to_excel(Path.joinpath(wdir_2025, 'TOC.xlsx'),index=False)
```

```python
map_data_init['Точка'] = map_data_init['Точка'].apply(lambda x: rename(x))
toc_data_init.to_excel(Path.joinpath(wdir_2025, 'Карты.xlsx'),index=False)
```
