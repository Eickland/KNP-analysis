```python
from pathlib import Path
import pandas as pd
```

```python
wdir_2025 = Path.cwd().parent / 'Initial data' / '2025'
toc_data_init = pd.read_excel(Path.joinpath(wdir_2025, 'TOC.xlsx')) # pyright: ignore[reportCallIssue]
map_data_init = pd.read_excel(Path.joinpath(wdir_2025, 'Карты.xlsx'),sheet_name=None)
```

```python
def rename(row):
    
    name = row
    split_name = name.split(sep='-')
    
    if 'B' in split_name[0]:
        return 'ADOM-SLB-'+ name
    
    elif 'K' in split_name[0] or 'К' in split_name[0]:
        return 'ADOM-SL'+ name[1:]
    
    else:
        return 'ADOM-SL'+ name
```

```python
map_data_list = []

for name,data in map_data_init.items():
    
    map_data_list.append(data)

map_data = pd.concat(map_data_list)
map_data.rename(columns={'Точка':'Sample','Широта':'lattitude','Долгота':'lontitude','pH':'pH','eH, mV':'eH'},inplace=True)
map_data.Sample = map_data.Sample.apply(lambda x: rename(x))
map_data
```

```python
map_data.to_excel(Path.joinpath(wdir_2025, 'GeoData.xlsx'),index=False)
```
