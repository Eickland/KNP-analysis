```python
import pandas as pd
from pathlib import Path
import HumSpectra.utilits as ut

date = '2024'
wdir = Path.cwd().parent / 'Initial data'/date

fluo_paths = wdir / 'Fluorescence'
uv_paths = wdir/'UV'
carbon_path = str(Path.joinpath(wdir,'C_org.xlsx'))
geodata_path = str(Path.joinpath(wdir,'GeoData.xlsx'))
```

```python
def rename_sample_names(old_name):

    split_dash_name = old_name.split(sep='-')
    split_whitespace_name = old_name.split(sep=' ')
    new_name = 'Error'
    
    if 'LG' == split_whitespace_name[0]:
        date = split_whitespace_name[2][:2]
        
        if 'LST' == split_whitespace_name[1]:
            new_name = split_whitespace_name[0] + '-' + split_whitespace_name[1] + '-' + date
            
        else:
            numbers_part = split_whitespace_name[1].split(sep='.')
            left_number = numbers_part[0]
            right_number = numbers_part[1]
            new_name = split_whitespace_name[0] + '-' + left_number + '-' + right_number + '-' + date
    
    elif 'LST' == split_whitespace_name[0]:
        
        if 'Rabochiy' == split_whitespace_name[1]:
            
            date = split_whitespace_name[-1][:2]
            new_name = split_whitespace_name[0] + '-' + 'Rab' + '-' + date
            
        else:
            
            new_name = ''
            
            for element in split_whitespace_name:
                new_name += element + '-'
                
            new_name = new_name[:-1]
                
            
    else:
        new_name =  old_name
        
    return new_name
```

```python
carbon_data = pd.read_excel(carbon_path)
geodata_data = pd.read_excel(geodata_path)
```

```python
for windows_path in uv_paths.rglob('*.csv'):
    path = str(windows_path)
    old_name = ut.extract_name_from_path(path)
    new_name = rename_sample_names(old_name)
    data = pd.read_csv(path)
    windows_path.unlink()
    data.to_csv(Path.joinpath(uv_paths,f'{new_name}.csv'),index=False)
```

```python
carbon_data.Sample = carbon_data.Sample.apply(lambda x: rename_sample_names(x))
geodata_data.Sample = geodata_data.Sample.apply(lambda x: rename_sample_names(x))
```

```python
carbon_data.to_excel(carbon_path,index=False)
geodata_data.to_excel(geodata_path,index=False)
```
