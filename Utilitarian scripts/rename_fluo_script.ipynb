```python
import pandas as pd
from pathlib import Path
```

```python
wdir = Path.cwd().parent
fluo_paths = Path.cwd().parent/ 'Initial data'/ '2025' / 'Fluorescence'
```

```python
def rename(row):
    
    name = row
    split_name = name.split(sep='-')
    
    if 'B' in split_name[0]:

        if 'BO' in split_name[0]:
            return 'ADOM-SLB-'+ name
        
        else:
            return 'ADOM-SLB-'+ name[1:]
    
    elif 'K' in split_name[0] or 'К' in split_name[0]:
        return 'ADOM-SL'+ name[1:]
    
    else:
        return 'ADOM-SL'+ name
```

```python
for windows_path in fluo_paths.rglob('*.csv'):
    path = str(windows_path)
    name = rename(path.split(sep="\\")[-1][:-4])
    print(path.split(sep="\\")[-1][:-4], name)
    df = pd.read_csv(path)
    windows_path.unlink()
    
    df.to_csv(str(windows_path.parent) +"\\"+ name+".csv")
```
